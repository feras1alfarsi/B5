### 1. Antibiotic use in children

The bar plot and the pie chart below show the distribution of pre-existing medical conditions of children involved in a study on the optimal duration of antibiotic use in treatment of tracheitis, which is an upper respiratory infection.

![](../../assets/viz_exercise_1.png)

a. What features are apparent in the bar plot but not in the pie chart?

b. What features are apparent in the pie chart but not in the bar plot?

c. Which graph would you prefer to use for displaying these categorical data?

Features apparent in the bar plot but not in the pie chart
Precise Values: The x-axis allows you to read specific numerical values (e.g., you can see "Prematurity" is exactly 33, whereas in the pie chart you can only estimate its size).

Easy Comparison of Similar Categories: It is much easier to see the small difference between "Neuromuscular" and "Trauma" because the bars are aligned on a common baseline.

Rank Order Clarity: The vertical list makes it immediately obvious which category is the 1st, 2nd, or 3rd most frequent without having to compare angles or areas.

Features apparent in the pie chart but not in the bar plot
Part-to-Whole Relationship: The pie chart explicitly shows how each category contributes to the 100% total. You can instantly see that "Prematurity" makes up roughly one-third of the total dataset.

Cumulative Proportions: By looking at segments together (like the top three categories), you can quickly gauge that they represent more than half of the total cases.

Which graph is preferred?
For this specific dataset, the Bar Plot is generally preferred.

Why?

Readability: Humans are much better at judging differences in length (bars) than differences in area or angles (pie slices).

Labeling: The bar plot keeps the labels ("Condition") right next to the data, whereas the pie chart requires the reader to look back and forth at a legend, which increases "cognitive load."

Small Categories: The categories at the bottom (Gastrointestinal and Immunocompromised) are very difficult to distinguish or measure in the pie chart but are clearly visible in the bar plot.

### 2. Black Lives Matter.

A Washington Post-Schar School poll conducted in the United States in June 2020, among a random national sample of 1,006 adults, asked respondents whether they support or oppose protests following George Floyd’s killing that have taken place in cities across the US. The survey also collected information on the age of the respondents. (Washington Post 2020) The results are summarized in the stacked bar plot below.

![](../../assets/viz_exercise_3.png)

a. Based on the stacked bar plot, do views on the protests and age appear to be associated? Explain your reasoning.

b. Conjecture other possible variables that might explain the potential association between these two variables.

Based on the stacked bar plot provided, yes, there appears to be a clear association between age and views on the protests.

Reasoning
The association is visible through the changing proportions of the "Opinion" categories across the different age groups:

Positive Correlation with Age and Opposition: As age increases, the proportion of respondents who "Strongly oppose" (the bright orange segment) grows significantly. In the 18-29 group, it is a very small sliver, but in the 65+ group, it covers nearly 25% of the total.

Negative Correlation with Age and Support: Conversely, the youngest demographic (18-29) shows the highest combined support. The "Strongly support" (darkest blue) and "Somewhat support" (lighter blue) segments shrink as you move up the y-axis toward the older age brackets.

Shift in Plurality: In the 18-29 group, "Strongly support" is the dominant opinion (occupying more than 50% of the bar). In the 65+ group, that dominant position is significantly diminished by the combined "Oppose" categories.

Conjectured Variables (Confounding Variables)
While the graph shows an association with age, other underlying factors might actually be driving this trend. Possible variables include:

Political Affiliation: Political leaning is often correlated with both age and views on social movements. Older generations might lean toward more conservative viewpoints, while younger generations may lean more progressive.

Media Consumption Habits: Different age groups consume news from different sources (e.g., TikTok/Social Media vs. Cable News). The framing of the protests in these different media bubbles could heavily influence opinion.

Socioeconomic Status: Older individuals are more likely to be homeowners or established in their careers, which can lead to a greater emphasis on social stability and "law and order" compared to younger individuals who may feel more impacted by the issues being protested.

Personal Experience/History: Older generations may compare current protests to historical events they lived through, while younger generations may view the protests as a necessary response to modern systemic issues.

### 3. Heart transplant data display.

The Stanford University Heart Transplant Study was conducted to determine whether an experimental heart transplant program increased lifespan. Each patient entering the program was officially designated a heart transplant candidate, meaning that they were gravely ill and might benefit from a new heart. Patients were randomly assigned into treatment and control groups. Patients in the treatment group received a transplant, and those in the control group did not. The visualizations below display two different versions of the study results.9 (Turnbull, Brown, and Hu 1974)

![](../../assets/viz_exercise_5.png)

a. Provide one aspect of the two group comparison that is easier to see from the stacked bar plot (left)?

b. Provide one aspect of the two group comparison that is easeir to see from the standardized bar plot (right)?

c. For the Heart Transplant Study which of those aspects would be more important to display? That is, which bar plot would be better as a data visualization?

Based on the image provided, here is an analysis of the two different ways to visualize the heart transplant study data:

Easier to see from the Stacked Bar Plot (Left)
The total sample size and the raw count of participants in each group are much easier to see.

You can immediately tell that the "treatment" group had significantly more participants (approx. 70) than the "control" group (approx. 35).

It also shows the absolute number of people who survived or passed away, which gives a sense of the scale of the study.

Easier to see from the Standardized Bar Plot (Right)
The relative survival rate (proportion) between the two groups is much easier to see.

By stretching both bars to 100% (1.00), you can compare the "alive" (blue) segments directly.

It becomes visually obvious that a much higher percentage of people survived in the treatment group compared to the control group, regardless of the fact that the group sizes were different.

Which is better for the Heart Transplant Study?
For a medical study like this, the Standardized Bar Plot (Right) is the better data visualization.

Reasoning:
In clinical trials, the primary goal is to determine the efficacy of a treatment. Because the two groups have different total numbers of patients, looking at raw counts (left plot) can be misleading or harder to interpret at a glance.

The standardized plot controls for the difference in group sizes and allows the viewer to focus on the probability of success. Since the "alive" segment is significantly larger in the treatment bar than the control bar, the visualization clearly communicates that the transplant (treatment) improved the chances of survival.

### 4. Meat consumption and life expectancy.

In data collected for You et al. (2022), total meat intake is associated with life expectancy (at birth) in 175 countries. c. How can you investigate the relationship between meat consumption and life expectancy in the presence of confounding variables (like income)?

Based on the density plots (also known as ridge plots) provided, here is an analysis of the relationship between meat consumption, life expectancy, and income.1. Do these graphs demonstrate an association?Yes, these graphs demonstrate an association, but it is an indirect one.Observation: In the top graph, High-income countries have the highest meat consumption (the blue peak is furthest to the right). In the bottom graph, those same High-income countries have the highest life expectancy (the blue peak is also furthest to the right).The Link: Conversely, Low-income countries have both the lowest meat consumption and the lowest life expectancy. Because both variables move in the same direction relative to income, we can infer they are associated: countries with low meat consumption do generally have lower life expectancy in this dataset.2. Explaining the Confounding VariableWhen you say the relationship is confounded by income level, you mean that income is a "hidden" third variable that influences both meat consumption and life expectancy simultaneously.Income $\rightarrow$ Meat: Wealthier countries can afford more expensive food sources like meat.Income $\rightarrow$ Life Expectancy: Wealthier countries generally have better healthcare systems, cleaner water, and better infrastructure, which directly increase life expectancy.Even if meat had zero biological effect on how long someone lives, you would still see them "associated" on a graph simply because rich countries happen to have more of both.3. Investigating in the presence of confoundingTo find the "true" relationship between meat and life expectancy while ignoring the "noise" of income, you can use the following methods:Stratification: Analyze the relationship within each income group separately. For example, look only at "Middle-income" countries and see if those who eat more meat within that specific group live longer. If the trend disappears when you look at one income level at a time, income was the true driver.Multivariable Regression: Use a statistical model where you "control" for income. This mathematically holds income constant to see if meat consumption still has a significant predictive effect on life expectancy.Scatter Plot with Color Coding: Create a scatter plot with Meat Consumption on the $x$-axis and Life Expectancy on the $y$-axis, then color the dots by Income Level. This allows you to visually check if the clusters for each income level show an internal trend or if the trend only exists between the clusters.
