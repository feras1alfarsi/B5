Here's a description of each column in the dataset:

- `total_bill`: The total bill amount, including the cost of food and drinks.
- `tip`: The tip amount given by the customer.
- `sex`: The gender of the customer (e.g., Male or Female).
- `smoker`: Whether the customer is a smoker or not (e.g., Yes or No).
- `day`: The day of the week when the transaction occurred (e.g., Sun, Sat, Thu, etc.).
- `time`: The time of day when the transaction occurred, typically categorized as Lunch or Dinner.
- `size`: The size of the party or group of customers.

**Your Task**: is to accept or reject the following hypothesis using statistical testing:

- Hypothesis $H_1$: smoking is associated with time of visit
- Hypothesis $H_2$: the bigger the group the higher the tip
- Hypothesis $H_3$: group size is different based on the time of visit
- Hypothesis $H_4$: (... come up with a hypothesis of your own ...)
- Finally, analyze if size (party size) is a **confounder**. That is, does a larger party cause a higher tip, or simply a higher bill which then leads to a higher tip?

---

## Hypothesis Testing Analysis Plan

### H₁: Smoking is associated with time of visit
**Test**: Chi-square test of independence
- Variables: `smoker` (categorical) vs `time` (categorical)
- Null hypothesis: Smoking status and time of visit are independent

### H₂: The bigger the group the higher the tip
**Test**: Pearson correlation or linear regression
- Variables: `size` (numeric) vs `tip` (numeric)
- Null hypothesis: No correlation between group size and tip amount

### H₃: Group size is different based on the time of visit
**Test**: Independent samples t-test or ANOVA
- Variables: `size` (numeric) across `time` (categorical)
- Null hypothesis: Mean group size is the same for Lunch and Dinner

### H₄: Males tip higher than females
**Test**: Independent samples t-test
- Variables: `tip` (numeric) across `sex` (categorical)
- Null hypothesis: Mean tip amount is the same for males and females

### Confounder Analysis: Does size confound the relationship between total_bill and tip?
**Approach**: 
1. Compare correlation between `total_bill` and `tip` (unadjusted)
2. Compare correlation between `total_bill` and `tip` controlling for `size` (adjusted)
3. If the adjusted correlation is substantially weaker, `size` may be a confounder

This analysis will help determine if larger parties lead to higher tips directly, or if the effect is mediated through higher bills.
