```python
import os
try:
    # In Colab? read from userdata (secrets)
    from google.colab import userdata
    ON_COLAB = True
    os.environ["SDAIA"] = userdata.get("SDAIA")
except ImportError:
    # Load `.env` file (locally)
    from dotenv import load_dotenv
    load_dotenv(override=True)
```

```python
from langchain_openai import ChatOpenAI
```

```python
# Low Temperature (0)
model_nemotron3_nano_precise = ChatOpenAI(
    model="nvidia/nemotron-3-nano-30b-a3b:free",
    temperature=0,
    # OpenRouter instead of the default OpenAI API
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ.get("SDAIA"),
)

```

```python
# High Temperature (0.9)
model_nemotron3_nano_creative = ChatOpenAI(
    model="nvidia/nemotron-3-nano-30b-a3b:free",
    temperature=0.9,
    # OpenRouter instead of the default OpenAI API
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ.get("SDAIA"),
)
```

```python
message = model_nemotron3_nano_creative.invoke("tell me about Saudi Arabia")
```

```python
print(message.content)
```

```python
message = model_nemotron3_nano_precise.invoke("tell me about Saudi Arabia")
```

```python
print(message.content)
```

```python
from langchain.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)
```

```python
system_prompt = SystemMessage("You always answer with 10 concise words, no more.Format your answer as JSON")

user_prompt = HumanMessage("tell me about Jeddah")

messages = [
    system_prompt,
    user_prompt,
]
response = model_nemotron3_nano_precise.invoke(messages)
```

```python
print(response.content)
```
