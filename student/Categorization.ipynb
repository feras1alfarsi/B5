```python
from langchain_openai import ChatOpenAI
```

```python

model_nemotron3_nano_precise = ChatOpenAI(
    model="nvidia/nemotron-3-nano-30b-a3b:free",
    temperature=0,
    # OpenRouter instead of the default OpenAI API
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ.get("SDAIA"),
)

```

```python
from pydantic import BaseModel
from typing import Literal
from langchain_core.prompts import ChatPromptTemplate

#Schema
class Tags(BaseModel):
    tags: list[Literal["cars", "shopping", "sports", "study", "work"]]

#Chain
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a text classifier. Assign only the relevant tags from the allowed list. Return multiple tags if applicable."),
    ("human", "{text}")
])

chain = prompt | model_nemotron3_nano_precise.with_structured_output(Tags)

#Sentences
sentences = [
    "That restoration looks incredible; you have a real talent for mechanics.",
    "I found the perfect gift today! The staff was incredibly helpful.",
    "Great game today! Your teamwork was the key to that victory.",
    "Learning together helped me finally grasp these concepts. Thank you!",
]

#Run
for sentence in sentences:
    result = chain.invoke({"text": sentence})
    print(f"Text : {sentence}")
    print(f"Tags : {result.tags}\n")
```
