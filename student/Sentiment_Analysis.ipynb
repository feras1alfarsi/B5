```python
from langchain_openai import ChatOpenAI
```

```python

model_nemotron3_nano_precise = ChatOpenAI(
    model="nvidia/nemotron-3-nano-30b-a3b:free",
    temperature=0,
    # OpenRouter instead of the default OpenAI API
    base_url="https://openrouter.ai/api/v1",
    api_key=os.environ.get("SDAIA"),
)

```

```python
from pydantic import BaseModel
from typing import Literal

class Sentiment(BaseModel):
    sentiment: Literal["positive", "neutral", "negative"]

chain = model_nemotron3_nano_precise.with_structured_output(Sentiment)
result = chain.invoke("The storm caused damage!")
print(result.sentiment)  # "positive"
```
