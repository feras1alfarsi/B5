### Step 1. Import the necessary libraries

```python
import pandas as pd
from pathlib import Path
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/02_Filtering_%26_Sorting/Euro12/Euro_2012_stats_TEAM.csv). 

### Step 3. Assign it to a variable called euro12.

```python
chipo = "https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/02_Filtering_%26_Sorting/Euro12/Euro_2012_stats_TEAM.csv"
chipo = pd.read_csv(chipo)
```

### Step 4. Select only the Goal column.

```python
goals = chipo['Goals']
goals
```

### Step 5. How many team participated in the Euro2012?

```python
num_teams = len(chipo)
print(f"Number of teams that participated in Euro2012: {num_teams}")
```

### Step 6. What is the number of columns in the dataset?

```python
num_columns = len(chipo.columns)
print(f"Number of columns in the dataset: {num_columns}")
```

### Step 7. View only the columns Team, Yellow Cards and Red Cards and assign them to a dataframe called discipline

```python
discipline = chipo[['Team', 'Yellow Cards', 'Red Cards']]
discipline
```

### Step 8. Sort the teams by Red Cards, then to Yellow Cards

```python
chipo[['Team', 'Red Cards','Yellow Cards']]
```

### Step 9. Calculate the mean Yellow Cards given per Team

```python
Yellow = chipo['Yellow Cards'].mean()
Yellow
```

### Step 10. Filter teams that scored more than 6 goals

```python
scored_more = chipo[chipo['Goals'] > 6]
scored_more
```

### Step 11. Select the teams that start with G

```python
teams_G = chipo[chipo['Team'].str.startswith('G')]
teams_G
```

### Step 12. Select the first 7 columns

```python
first_7_columns = chipo.iloc[:, :7]
first_7_columns
```

### Step 13. Select all columns except the last 3.

```python
last_3_columns = chipo.iloc[:, -3:]
last_3_columns
```

### Step 14. Present only the Shooting Accuracy from England, Italy and Russia

```python
countries = ['England', 'Italy', 'Russia']
result = chipo[chipo['Team'].isin(countries)][['Team', 'Shooting Accuracy']]
result
```
