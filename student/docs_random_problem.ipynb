````python
import random
import string
````

````python


def generate_password(length: int, min_numbers: int = 1, min_special: int = 1) -> str:
    
    upper = string.ascii_uppercase
    lower = string.ascii_lowercase
    digits = string.digits
    special = "!@#$%^&*"
    
    
    password = [
        random.choice(upper),
        random.choice(lower)
    ]
    
    
    for _ in range(min_numbers):
        password.append(random.choice(digits))
        
    for _ in range(min_special):
        password.append(random.choice(special))

    all_possible = upper + lower + digits + special
    
    while len(password) < length:
        password.append(random.choice(all_possible))

    random.shuffle(password)
    
   
    return "".join(password)


print(generate_password(10, min_numbers=2, min_special=2))
````

````python
import random

def generate_passphrase(num_words: int = 4) -> str:
    
    word_list = [
        "apple", "cloud", "river", "bridge", "mountain", 
        "falcon", "sunset", "rocket", "garden", "planet",
        "coffee", "forest", "window", "ocean", "silver"
    ]

    chosen_words = random.choices(word_list, k=num_words)
    return "-".join(chosen_words)

print(f"Your Passphrase: {generate_passphrase(4)}")
````
