```python
from langchain.messages import (
    SystemMessage,
    HumanMessage,
    AIMessage
)
```

```python
system_prompt = SystemMessage("""
You are a math assistant. You have access to these tools:

def add (a , b):
  result = a + b
  return result

def subtracit  (a , b):
  result = a - b
  return result

def multiply  (a , b):
  result = a * b
  return result

def divide  (a , b):
  result = a / b
  return result

Return the tool name and arguments required to solve the user's request.
""")

user_prompt = HumanMessage("what is two plus 5")

messages = [
    system_prompt,
    user_prompt,
]
response = model_nemotron3_nano_precise.invoke(messages)
```

```python
from pydantic import BaseModel, Field
from typing import Optional
class ToolCall(BaseModel):
    tool_name: str = Field(description="The name of the function to use")
    arguments: Dict[str, Any] = Field(description="The parameters to pass")
```
