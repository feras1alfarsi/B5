### Step 11. Print only the occupation column

```python
# Load the occupations dataset
users_id = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/u.user', sep='|')

# Now you can print the column
print(users_id['occupation'])
```

### Step 12. How many different occupations are in this dataset?

```python

num_occupations = users_id['occupation'].nunique()

print(f"There are {num_occupations} different occupations in the dataset.")
```

### Step 13. What is the most frequent occupation?

```python
most_occupation = users_id['occupation'].value_counts().idxmax()
total_count = users_id['occupation'].value_counts().max()

print(f"The most common occupation is: {most_occupation}")
print(f"It appeared {total_count} times.")
```

### Step 18. What is the age with least occurrence?

```python
age_counts = users_id['age'].value_counts()
least_common_age = age_counts.idxmin()
least_common_age_count = age_counts.min()
print(f"The least common age is: {least_common_age}")
print(f"It appeared {least_common_age_count} times.")   
```
