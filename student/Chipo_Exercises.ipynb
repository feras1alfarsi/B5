### Step 9. Which was the most-ordered item? 

```python
# This counts every occurrence of each item and sorts them automatically
most_ordered = chipo['item_name'].value_counts().idxmax()
quantity = chipo ['item_name'].value_counts().max()

print(f"The most-ordered item is: {most_ordered}")
print(f"It was ordered {quantity} times.")
```

### Step 10. For the most-ordered item, how many items were ordered?

```python

item_counts = chipo.groupby('item_name')['quantity'].sum()
most_ordered_count = item_counts.max()

print(f"The most-ordered item was ordered {most_ordered_count} times.")
```

### Step 11. What was the most ordered item in the choice_description column?

```python
most_ordered_choice = chipo['choice_description'].value_counts().idxmax()
total_count = chipo['choice_description'].value_counts().max()

print(f"The most ordered choice description is: {most_ordered_choice}")
print(f"It appeared {total_count} times.")
```

### Step 12. How many items were orderd in total?
