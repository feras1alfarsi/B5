### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv). 

### Step 3. Assign it to a variable called chipo.

```python
chipo = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(chipo, sep="\t")
```

```python
print(chipo.columns)
```

### Step 4. How many products cost more than $10.00?

```python
# 1. Remove the '$' and convert 'item_price' to a decimal (float)
chipo['item_price'] = chipo['item_price'].apply(lambda x: float(x.replace('$', '')))

# 2. Now the division will work because both sides are numbers!
chipo['unit_price'] = chipo['item_price'] / chipo['quantity']

# 3. Filter for products > $10 and count unique names
products_over_10 = chipo[chipo['unit_price'] > 10.00]['item_name'].nunique()

print(f"Number of unique products over $10: {products_over_10}")
```

```python
# 2. Filter for unit prices > 10.00 and drop duplicates (to count unique products)
products_over_10 = chipo[chipo['unit_price'] > 10.00]['item_name'].nunique()

print(f"There are {products_over_10} unique products that cost more than $10.00.")
```

### Step 5. What is the price of each item? 
###### print a data frame with only three columns item_name choice_description and product_price

```python
# Create a dataframe with only item_name, choice_description, and unit_price
price_df = chipo[['item_name', 'choice_description', 'unit_price']].drop_duplicates()

price_df
```

### Step 6. Sort by the name of the item

```python
# Sort price_df by item_name
price_df_sorted = price_df.sort_values('item_name')

price_df_sorted
```

### Step 7. What was the quantity of the most expensive item ordered?

```python
# Find the most expensive item (highest unit_price)
most_expensive_item = chipo.loc[chipo['unit_price'].idxmax()]

# Get the quantity of the most expensive item ordered
quantity_most_expensive = most_expensive_item['quantity']

print(f"The most expensive item '{most_expensive_item['item_name']}' with unit price ${most_expensive_item['unit_price']:.2f} was ordered with a quantity of {quantity_most_expensive}.")
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad_bowl_orders = chipo[chipo['item_name'] == 'Veggie Salad Bowl']
times_ordered = len(veggie_salad_bowl_orders)
print(f"Veggie Salad Bowl was ordered {times_ordered} times.")
```

### Step 9. How many times did someone order more than one Canned Soda?

```python
# Filter for orders of Canned Soda with quantity > 1
canned_soda_more_than_one = chipo[(chipo['item_name'] == 'Canned Soda') & (chipo['quantity'] > 1)]

# Count the number of such orders
count = len(canned_soda_more_than_one)



print(f"Number of times someone ordered more than one Canned Soda: {count}")
```
